## 1. 토큰 받기

[→ §300 강의노트](../notes/FD-14-300__account-and-trading.md)

```python
# 📌 §300 토큰 받기
token = get_token()
print(token[:8])
# 예상 출력: eyJ0eXAi
```

## 2. 계좌 확인

[→ §300 강의노트](../notes/FD-14-300__account-and-trading.md)

보유 종목 목록을 읽는다. 읽기 전용이라 주문은 나가지 않는다.


```python
# 📌 §300 계좌 확인
balance = fetch_balance(token)
print(len(balance))
# 예상 출력: 0
```

## 3. 금액 확인

[→ §300 강의노트](../notes/FD-14-300__account-and-trading.md)

현금, 주식 평가금액, 총 평가금액만 확인한다.


```python
# 📌 §300 금액 확인
summary = fetch_account_summary(token)
print(summary)
# 예상 출력: {'cash': '1,000,000원', ...}
```

[→ §300 강의노트](../notes/FD-14-300__account-and-trading.md)

## 5. 모의 주문해보기

[→ §300 강의노트](../notes/FD-14-300__account-and-trading.md)

```python
# 📌 §300 모의 주문
confirm_mock_order = True
if confirm_mock_order:
    response = place_order(token, ticker, qty)
    print(response.get("msg1", response))
else:
    print("교수 확인 후 confirm_mock_order를 True로 바꾼다")
# 예상 출력: 교수 확인 후 confirm_mock_order를 True로 바꾼다
```
